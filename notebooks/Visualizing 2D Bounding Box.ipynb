## Visualizing 2D bounding box attributes
To run this notebook you first have to generate the `toy_dataset` which can be generated by executing `make toy_dataset`

```python
import json  # Used to read the dataset file
import cv2  # Used to read the images
import matplotlib.pyplot as plt  # Used to plot
import matplotlib.patches as patches  # Used to draw the bb
%matplotlib inline
```

### Read the dataset file

```python
# Read the dataset index file
with open('../data/toy_dataset/blendgen_dataset.json') as json_file:
    data = json.load(json_file)
```

### Read the image file

```python
# The frame of the dataset we will use for visualization
FRAME_NUM = 5

# Plot the color image
color = cv2.cvtColor(cv2.imread('../data/toy_dataset/color/Image{:0>4}.png'.format(FRAME_NUM)), cv2.COLOR_RGB2BGR)
plt.imshow(color)
plt.show()
```

### Draw screen-space bounding box over image

```python
fig = plt.figure(figsize=(16,9))
ax = fig.add_subplot(111)
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')

# Draw the color image
plt.imshow(color)

"""
Grab the bounding box for a given frame. The format we saved it in
in the toy_dataset example is [width, height, x, y]
"""
bounding_box = data['data'][FRAME_NUM]['attributes'][0]['bounding_box']
rect = patches.Rectangle((bounding_box[0],
                          bounding_box[1]),
                         bounding_box[2],
                         bounding_box[3],
                         linewidth=1,
                         edgecolor='r',
                         facecolor='none')
ax.add_patch(rect)    
plt.show()
```
